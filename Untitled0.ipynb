```python
# ============================================================
# YOLO WEED / CROP DETECTION - GOOGLE COLAB TRAINING
# ============================================================

# 1. Install Ultralytics YOLO
!pip install ultralytics -q

import torch

print("CUDA available:", torch.cuda.is_available())
print("GPU count:", torch.cuda.device_count())

if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))
# 2. Import libraries
from google.colab import drive
from ultralytics import YOLO
import os


# 3. Mount Google Drive
drive.mount('/content/drive')


# ============================================================
# 4. SET DATASET PATH
# ============================================================

# Your data.yaml file
yaml_path = "/content/drive/MyDrive/weed_yolo_prepared/data.yaml"

# Check that the YAML file exists
if not os.path.exists(yaml_path):
    raise FileNotFoundError(f"data.yaml not found: {yaml_path}")

print("data.yaml found!")
print(yaml_path)


# ============================================================
# 5. CHECK DATASET FOLDERS
# ============================================================

dataset_path = "/content/drive/MyDrive/weed_yolo_prepared/dataset"

train_images = os.path.join(dataset_path, "images/train")
val_images   = os.path.join(dataset_path, "images/val")

print("\nChecking dataset...")

if os.path.exists(train_images):
    print("Training images folder: OK")
else:
    print("ERROR: Training images folder not found!")
    print(train_images)

if os.path.exists(val_images):
    print("Validation images folder: OK")
else:
    print("ERROR: Validation images folder not found!")
    print(val_images)


# ============================================================
# 6. LOAD YOLO MODEL
# ============================================================

# YOLOv8 Nano model
# Small and suitable for initial training/testing
model = YOLO("yolov8n.pt")


# ============================================================
# 7. TRAIN THE MODEL
# ============================================================

results = model.train(
    data=yaml_path,

    # Number of training epochs
    epochs=50,

    # Image size
    imgsz=640,

    # Batch size
    batch=16,

    # Save training results directly to Google Drive
    project="/content/drive/MyDrive/weed_yolo_training",

    # Training folder name
    name="weed_detection",

    # Use GPU if available
    device=0,

    # Save model checkpoints
    save=True,

    # Display training progress
    verbose=True
)


# ============================================================
# 8. TRAINED MODEL LOCATION
# ============================================================

best_model = (
    "/content/drive/MyDrive/weed_yolo_training/"
    "weed_detection/weights/best.pt"
)

last_model = (
    "/content/drive/MyDrive/weed_yolo_training/"
    "weed_detection/weights/last.pt"
)

print("\n==========================================")
print("TRAINING COMPLETED")
print("==========================================")

print("\nBest model:")
print(best_model)

print("\nLast model:")
print(last_model)


# ============================================================
# 9. CHECK IF BEST MODEL EXISTS
# ============================================================

if os.path.exists(best_model):
    print("\nSUCCESS: best.pt was created!")
else:
    print("\nWARNING: best.pt was not found.")


# ============================================================
# 10. VALIDATE THE TRAINED MODEL
# ============================================================

print("\nRunning validation...")

trained_model = YOLO(best_model)

metrics = trained_model.val(
    data=yaml_path,
    imgsz=640,
    batch=16,
    device=0
)

print("\n==========================================")
print("VALIDATION COMPLETED")
print("==========================================")
```
